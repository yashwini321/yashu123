```python
import pandas as pd
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.read_csv('iris (1).csv')
print(df)
print("missing values in each columns:",df.isnull().sum())
mean=df['sepal_length'].mean()
print(mean)
df=df['sepal_length'].fillna(mean)
print(df)
print("missing values after replacement:",df.isnull().sum())
print("5 rows:",df.head())
```
