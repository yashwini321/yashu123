```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
Cars_df=pd.read_csv('/content/Cars93 (1).csv')
print(Cars_df)
fig,ax=plt.subplots(figsize=(10,4))
ax.scatter(Cars_df['Horsepower'],Cars_df['Price'])
ax.set_xlabel('Horesepower')
ax.set_ylabel('Price')
plt.title("Scatterplot")
plt.show()
Cars_df['Horsepower'].plot(kind='box')
plt.show()
q1=Cars_df['Horsepower'].quantile(0.25)
q3=Cars_df['Horsepower'].quantile(0.75)
iqr=q3-q1
upper_bound=q3+(1.5*iqr)
lower_bound=q1-(1.5*iqr)
Horsepower_arr=Cars_df["Horsepower"]
outliers=Horsepower_arr[(Horsepower_arr<=lower_bound)|(Horsepower_arr>=upper_bound)]
print('The following are the outliers in the boxplot of Horsepower:\n',outliers)
Cars_df['Price'].plot(kind='hist')
plt.show()
upperIndex=Cars_df[Cars_df['Horsepower']>upper_bound].index
Cars_df.drop(upperIndex,inplace=True)
lowerIndex=Cars_df[Cars_df['Horsepower']<lower_bound].index
Cars_df.drop(lowerIndex,inplace=True)
print(Cars_df.info())
Cars_df=pd.read_csv('/content/Cars93 (1).csv')
Price_arr=Cars_df['Price']
upper_cap=np.percentile(Price_arr,1)
lower_cap=np.percentile(Price_arr,99)
outliers=Price_arr[(Price_arr<upper_cap)|(Price_arr>lower_cap)]
print("the following are the outliers in the boxplot of Price:\n",outliers)
for i in Cars_df['Price']:
  if i<lower_bound:
    Cars_df['Price']=Cars_df['Price'].replace(i,lower_cap)
  elif i>upper_bound:
    Cars_df['Price']=Cars_df['Price'].replace(i,upper_cap)
Cars_df.info()
Cars_df=pd.read_csv('/content/Cars93 (1).csv')
m=np.mean(Cars_df['Horsepower'])
print('mean:',m)
for i in Cars_df['Horsepower']:
   if i<lower_bound or i>upper_bound:
    Cars_df['Horsepower']=Cars_df['Horsepower'].replace(i,m)
```
